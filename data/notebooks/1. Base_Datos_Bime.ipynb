- Eliminación de espacios innecesarios.

Este proceso asegura homogeneidad en la variable `localidad` y evita duplicidades causadas por diferencias ortográficas.

```python
import unicodedata

def limpiar_localidad(x):
    if pd.isna(x):
        return x
    x = str(x)

    # intentar corregir mala codificación
    try:
        x = x.encode("latin1").decode("utf-8")
    except:
        pass  # si ya está bien, no hace nada

    # quitar tildes
    x = unicodedata.normalize("NFKD", x).encode("ascii", "ignore").decode("ascii")

    # pasar a mayúsculas
    x = x.upper().strip()

    return x

df["localidad"] = df["localidad"].apply(limpiar_localidad)

df["localidad"].unique()
```
